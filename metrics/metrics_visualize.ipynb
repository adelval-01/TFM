```python
import numpy as np
from scipy.io import savemat

data = np.load("/home/adelval/BTS/TFM/metrics/all_audio_latencies.npz")

# Ensure values are numeric arrays
mat_dict = {}
for key in data.files:
    arr = data[key]
    if isinstance(arr, np.ndarray):
        mat_dict[key] = arr.astype(np.float64)  # Ensures MATLAB compatibility

savemat("/home/adelval/BTS/TFM/metrics/all_audio_latencies.mat", mat_dict)
```
